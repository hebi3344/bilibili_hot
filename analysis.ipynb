```python
import re
import pandas as pd
import sqlite3
import os
```

```python
DB_DIR = os.path.join(os.path.expanduser("~"), "job_data")
DB_PATH = os.path.join(DB_DIR, 'video_data.db')
conn = sqlite3.connect(DB_PATH)
df = pd.read_sql_query("SELECT * FROM video", conn)
conn.close()
```

```python
#计算总播放量
def cul_v(df):
    df = str(df).strip()
    pattern = r'([\d.]+)'
    match = re.search(pattern,df)
    A_count = float(match.group(1))
    if '亿' in df:
        A_count *= 100000000
    elif '万' in df:
        A_count *= 10000
    return int(A_count)
df['播放量_数字'] = df['播放量'].apply(cul_v)
print(df[['播放量','播放量_数字']])
total_v_count = df['播放量_数字'].sum()
print(f'总播放量:{total_v_count}')
```
